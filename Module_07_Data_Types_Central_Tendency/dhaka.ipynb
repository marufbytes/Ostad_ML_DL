```python
import numpy as np
import pandas as pd
```

```python
numbers = [2,4,6,8,10]
```

```python
np_variance = np.var(numbers)
print(f"Variance by Nupy: {np_variance}")
```

```python
np_std = np.std(numbers)
print(f"Standard Deviation by Numpy: {np_std}")
```

```python
num_series = pd.Series(numbers)
pd_variance = num_series.var()
print(f"Variance by Panda: {pd_variance}")
```

```python
pd_std = num_series.std()
print(f"Standard Deviation by Numpy: {pd_std}")
```
