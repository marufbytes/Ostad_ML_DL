```python
fig, ax = plt.subplots(figsize=(6, 4))
sns.scatterplot(data=tips, x='total_bill', y='tip', ax=ax) 
ax.set_title("Bill vs Tip")
plt.show()
```

```python

sns.lmplot(data=tips, x='total_bill', y='tip', height=4, aspect=2)

plt.suptitle("Scatterplot of Total Bill vs Tip")
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(data=tips, x='total_bill', bins=10, color='blue', ax=axes[0])
axes[0].set_title("Histogram total bill")

sns.kdeplot(data=tips, x='total_bill', color='blue', ax=axes[1])
axes[1].set_title("Kernel Density Estimation of Total bill")

sns.histplot(data=tips, x='total_bill', color='green', kde=True, ax=axes[2])
axes[2].set_title("Histogram+Kernel Density Estimation of Total bill")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from ipywidgets import Dropdown, interact

tips = sns.load_dataset("tip" if "tips" not in locals() else "tips")

numeric_cols = tips.select_dtypes(include=np.number).columns.tolist()
category_cols = ["sex", "smoker", "day", "time"]


def interactive_scatter(x_col, y_col, hue_col):
  plt.figure(figsize=(12,5))
  sns.scatterplot(
      data=tips, x=x_col, y=y_col, hue=hue_col, palette="Set1"
  )
  plt.title(f"{y_col} vs {x_col} by {hue_col}")
  plt.tight_layout()
  plt.show()


interact(
    interactive_scatter,
    x_col=Dropdown(options=numeric_cols, value="total_bill"),
    y_col=Dropdown(
        options=numeric_cols, value="tip"
    ),  
    hue_col=Dropdown(options=category_cols, value="time"),
)
```
