```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('2026-08-02T04-22-33.232Z-employee_dataset.csv')
df
```

```python
df.info()
```

```python
plt.figure(figsize=(10,5))
plt.scatter(
    df['Experience'],
    df['Salary'],
    s = df['Projects_Completed']*80, 
    alpha = 0.5
)
plt.title("Experience vs Salary (Bubble = Project Completed)")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,5))
scatter = plt.scatter(
    df['Salary'],
    df['Performance_Score'],
    c=df['Training_Hours'],
    cmap='viridis'
)
plt.colorbar(scatter, label='Training Hours')  
plt.title("Salary vs Performance")
plt.xlabel("Salary")
plt.ylabel("Performance")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Aroj sundor style er jonno
plt.style.use('seaborn-v0_8-whitegrid')

plt.figure(figsize=(10, 5))
scatter = plt.scatter(
    df['Experience'],
    df['Salary'],
    s=df['Projects_Completed'] * 80,
    c=df['Projects_Completed'],  # Projects er upor base kore color change hobe
    cmap='viridis',              # Attractive colormap
    alpha=0.7,
    edgecolors='black',          # Bubble er charpashe black border
    linewidth=0.8
)

plt.title("Experience vs Salary (Bubble size & color = Projects Completed)", fontsize=14, fontweight='bold')
plt.xlabel("Experience", fontsize=12)
plt.ylabel("Salary", fontsize=12)

# Side e color bar add korbe
plt.colorbar(scatter, label="Projects Completed")
plt.show()
```
