```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,4,5,6]
y=[2,4,6,8,9]

plt.plot(x,y)

plt.plot(x,y, color = "red", marker = 'o',linestyle='--', linewidth=3)

plt.xlabel('Day')
plt.ylabel('Sales')
plt.title('Daily Sales')
plt.grid()


plt.show()
```

```python
products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [120, 180, 90, 150]
plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product vs Sales")
```

```python
days=[1,2,3,4,5]
sales=[100,120,130,140,160]
plt.plot(days,sales, color='green', marker="o")
plt.xlabel("Day")
plt.ylabel("Sales")
plt.title("Daily sales")
plt.grid()
plt.show()
```

```python
subjects = ["Python", "SQL", "ML", "AI"]
students = [40,25,34,63]
plt.pie(students, labels=subjects)
plt.show()
```
